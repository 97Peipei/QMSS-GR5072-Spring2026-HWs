## 5.Write a for loop that loops through each level of the categorical factor you chose in Q3, where each iteration of the for loop subsets your data for one group of the categorical variable at a time, and then fits the same regression model specified above in Q4 k times, where k refers to the number of unique categories. On each iteration of your for loop, store the model.summary() information as a new element in a list object. Once your for loop is done running, print the list. There should be k regressions which you estimated. Does the effect of LotArea on SalePrice vary within the different levels of the categorical factor you chose? What is another, more common way to answer a question like this (i.e., a question about moderation or effect heterogeneity)?

```python
summary_list = []
```

```python
# loop through each category in PavedDrive
for level in df["PavedDrive"].dropna().unique():
    
    subset = df[df["PavedDrive"] == level]
    
    x_sub = subset[["LotArea", "OverallQual", "OverallCond"]]
    y_sub = subset["SalePrice"]
    
    x_sub = sm.add_constant(x_sub)
    
    model_sub = sm.OLS(y_sub, x_sub).fit()
    
    # store summary
    summary_list.append(model_sub.summary())
    
    # store LotArea coefficient and standard error
    print(f"--- PavedDrive Level: {level} ---")
    print(f"LotArea Coef: {model_sub.params['LotArea']:.4f}")
    print(f"LotArea StdErr: {model_sub.bse['LotArea']:.4f}")
    print("\n")

# print all regression summaries
for s in summary_list:
    print(s)
    print("\n" + "="*100 + "\n")
```

Yes, the effect of `LotArea` on `SalePrice` appears to vary across the different levels of the categorical variable `PavedDrive`. 

From the subgroup regressions, the estimated coefficient of `LotArea` is lowest for homes with paved drive access (`Y`), higher for homes with no paved drive (`N`), and highest for homes with partial paved drive (`P`). This suggests that the relationship between lot size and sale price is not constant across groups.

However, these differences should be interpreted with caution. The standard errors differ across groups, and the sample size for some categories (especially `P`) is relatively small, which makes the estimates less precise. Therefore, the observed differences in coefficients may not necessarily indicate statistically significant differences.

A more common and formal way to answer this type of question is to estimate a single regression model that includes interaction terms between `LotArea` and the categorical variable `PavedDrive`. This allows us to directly test whether the effect of `LotArea` on `SalePrice` differs across groups, rather than comparing separate regressions.
