## 6.Use this variable to create two variables that are called "first_position" and "second_position". Hint: If a player plays two positions, then the "-" is used as a delimiter between their first and second positions. In these cases, you will have to split the position variable into two.

```python
df[['first_position', 'second_position']] = df['Position'].str.split('-', expand=True)

display(df[['Player', 'Position', 'first_position', 'second_position']].head())
display(df[['Player', 'Position', 'first_position', 'second_position']].tail())
```

## 7.Combine these two variables back into a single variable called "position_united". Make sure to separate the two positions by "-", as the original data field did. If there is no second position, then there will only be a single position. Again, this is similar to how the original variable was coded. Present a frequency table of "position_united".

```python
def combine_positions(row):
    if row['second_position'] is None or pd.isna(row['second_position']):
        return row['first_position']
    else:
        return f"{row['first_position']}-{row['second_position']}"
    
```

```python
df['position_united'] = df.apply(combine_positions, axis=1)

result_Q7 = df['position_united'].value_counts().reset_index()
result_Q7.columns = ['position_united', 'count']

display(result_Q7)
```

## 8. Create two new datasets.

a new dataset from the original dataset that includes all data except the age variable (be sure to give this dataset a new name).

add a new column to both datasets called mergeid that includes a sequence of numbers beginning with a 1 in the first row of the data and ending with the total number of rows in the last row of the data.

## 9.Join the two datasets from question (8) together to recreate the original dataset plus the new merge id.

```python
df_recreated = pd.merge(df_no_age, df_player_age, on='mergeid')

print("Recreated Dataset (Joined on mergeid) :")
display(df_recreated.head())
display(df_recreated.tail())
```

## 10.Subset the original dataset to the years 1995-1997. Group the data by team name and year and then summarize the average number of points per team. Arrange from most to least points. 

(Hint: Your answer should have a shape of 88 x 3, since we are looking at the average points scored over the 1995-1997 seasons. Finally, also note that the data for VAN and TOR are missing in 1995 as these teams did not exist in that year. So your answer could also be 90 x 3, if you include missing/empty data in these rows).
