## 1.Which NBA player scored the most points in a single season? What seasons/year was this?

## 2.Which NBA player scored the most points in 1991?

```python
df_1991 = df[df['Year'] == 1991]

top_scorer_1991 = df_1991[df_1991['PTS'] == df_1991['PTS'].max()]

display(top_scorer_1991[['Player','Year','PTS']])
```

## 3.Which player had the best free throw percentage ("FT%") from the year 2000 to the most recent year in the data? If there are multiple players who have tied for the best percentage, then sort on the column with player names and submit your answer as a pd.DataFrame object.

```python
df_Q3 = df[df['Year']>= 2000]

max_FT = df_Q3['FT%'].max()
result_Q3 = df_Q3[df_Q3['FT%'] == max_FT].sort_values(by='Player')

display(result_Q3[['Player', 'Year', 'FT%']].head())
display(result_Q3[['Player', 'Year', 'FT%']].tail())
```
