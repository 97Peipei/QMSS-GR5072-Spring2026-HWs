## 6.Use this variable to create two variables that are called "first_position" and "second_position". Hint: If a player plays two positions, then the "-" is used as a delimiter between their first and second positions. In these cases, you will have to split the position variable into two.

```python
df[['first_position', 'second_position']] = df['Position'].str.split('-', expand=True)

display(df[['Player', 'Position', 'first_position', 'second_position']].head())
display(df[['Player', 'Position', 'first_position', 'second_position']].tail())
```

## 7.Combine these two variables back into a single variable called "position_united". Make sure to separate the two positions by "-", as the original data field did. If there is no second position, then there will only be a single position. Again, this is similar to how the original variable was coded. Present a frequency table of "position_united".

```python
def combine_positions(row):
    if row['second_position'] is None or pd.isna(row['second_position']):
        return row['first_position']
    else:
        return f"{row['first_position']}-{row['second_position']}"
    
```

```python
df['position_united'] = df.apply(combine_positions, axis=1)

result_Q7 = df['position_united'].value_counts().reset_index()
result_Q7.columns = ['position_united', 'count']

display(result_Q7)
```
